# Candidate feature inventory and leakage notes
Decision-neutral inventory of sources, cutoff availability, and risks.

## Simple explanation of the table

Each row is one possible input to a future podium model. `candidate_feature` is the proposed variable, `source` says where it comes from, `available_at_cutoff` says whether it is known before the race starts, and `information_time` says when that information becomes available. `risk_or_note` records an important warning or definition. `Yes` means the feature is available before the prediction; `No` means it would reveal the race result or something that happened during the race and must not be used; `Where available` means coverage differs by season; and `Open` means the team still needs to decide whether to include it. For example, `grid` and `quali_pos` describe the current weekend, historical form features use only previous races, and `positionOrder`, `status`, `points`, `laps`, and pit-stop data are forbidden because they are known only after or during the race.

```python
import pandas as pd
inventory=pd.DataFrame([
['grid','results.grid','Yes','Current weekend','Penalties may separate it from qualifying position'],['pitlane_start','results.grid + laps','Yes','Current weekend','Separate pit-lane starts from non-starters'],['quali_pos','qualifying.position','Where available','Current weekend','Era-dependent coverage'],['q1_time','qualifying.q1','Where available','Current weekend','Structural missingness'],['q2_time','qualifying.q2','Where available','Current weekend','Unavailable in older formats'],['q3_time','qualifying.q3','Where available','Current weekend','Unavailable in older formats'],['driver_age','drivers.dob + races.date','Yes','Current weekend','Date parsing'],['home_race','drivers.nationality + circuits.country','Yes','Current weekend','Mapping and association-not-causation risk'],['driver_form_3/5/10','Prior results','Yes after history exists','Previous races','Must shift before rolling'],['constructor_form_3/5/10','Prior results','Yes after history exists','Previous races','Constructor identity/rebrands'],['circuit_history','Prior results at circuit','Yes after history exists','Previous races','Sparse history'],['driver_standing','Previous driver standings','Yes if lagged','Previous round','Current round leaks outcome'],['constructor_standing','Previous constructor standings','Yes if lagged','Previous round','Current round leaks outcome'],['mechanical_dnf_rate','Prior status taxonomy','Yes after history exists','Previous races','Taxonomy dependency'],['field_size','Race entry list','Yes with care','Current race setup','Define entrants carefully'],['positionOrder','results.positionOrder','No','Current race','Target/post-race leakage'],['status','results.statusId + status','No','Current race','Forbidden predictor'],['points/laps/fastestLap','results','No','Current race','Forbidden current-race predictors'],['lap_times/pit_stops','lap_times / pit_stops','No','During race','After cutoff'],['sprint_result','sprint_results','Open','Before Sunday where applicable','Historically inconsistent']],columns=['candidate_feature','source','available_at_cutoff','information_time','risk_or_note'])
display(inventory); print('Forbidden:',(inventory.available_at_cutoff=='No').sum())
```
